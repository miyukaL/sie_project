# Example usage of the sanity_check_resulty.csv file #

```python
import pandas as pd
```

```python
# count of files which are present And are not nan : 
files_exist = pd.read_csv("sanity_check_output/sanity_check_results.csv")
print(f"Total files expected: {len(files_exist)}")
print(f"Number of files found: {(files_exist['is_file_available'].sum())}")
print(f"Percentage of files found: {(files_exist['is_file_available'].mean() * 100):.2f}%")
```

```python
# count of only nan files : 
nan_csv = pd.read_csv("sanity_check_output/sanity_check_results.csv")
print(f"{nan_csv['all_nan'].sum()} files with only NaN")
print(f"Among files found : {(nan_csv['all_nan'].mean()*100):.2f}% are all NaN")
print(f"Percentage available nonNan hours on the year : {nan_csv['is_file_available'].sum()*(1-(nan_csv['all_nan'].mean()))/len(nan_csv)*100 :.2f}%")
```

```python
# percentage troncated files :
trunc_csv = pd.read_csv("sanity_check_output/sanity_check_results.csv")
print(f"{trunc_csv['is_truncated'].sum()} troncated files before the 07/15")
print(f"making {(trunc_csv['is_truncated'].mean()*100):.2f}% of the files")
```

```python
# isolated in time : 
one_time = pd.read_csv("sanity_check_output/sanity_check_results.csv")
print(f"{one_time['is_single_timestamp'].sum()} files with only one timestamps")
print(f"making {(one_time['is_single_timestamp'].mean()*100):.2f}% of the files")
```

```python
# isolated in time for range > 4 : 
one_time_r4 = pd.read_csv("sanity_check_output/sanity_check_results.csv")
print(f"{one_time_r4['is_single_timestamp_anfa'].sum()} files with only one timestamp for range > 4")
print(f"making {(one_time_r4['is_single_timestamp_anfa'].mean()*100):.2f}% of the files")
```

```python
# aberrant timestamps : 
weird_ts = pd.read_csv("sanity_check_output/sanity_check_results.csv")
flagged = weird_ts['n_weird_timesteps'].fillna(0) > 0
print(f"{flagged.sum()} files with aberrant timestamps (variations > 0.2s compared to 5 or 10s)")
print(f"making {(flagged.sum()/weird_ts['is_file_available'].sum()*100):.2f}% of the files")
```

```python
# near ground artefact :
near_ground = pd.read_csv("sanity_check_output/sanity_check_results.csv")
print(f"{near_ground['is_contaminated_nfa'].sum()} files contaminated by the near field artefact")
print(f"making {near_ground['is_contaminated_nfa'].mean()*100:.1f}% of the files")
```

```python
# aberrant Zea values :
aberrant_values = pd.read_csv("sanity_check_output/sanity_check_results.csv")
flagged = aberrant_values['n_aberrant_val'].fillna(0) > 0
print(f"{flagged.sum()} files with aberrant values of Zea")
print(f"making {flagged.sum()/aberrant_values['is_file_available'].sum()*100:.2f}% of the files")
```

```python
# coherence Zea vs Vel vs WIDTH :
coherent = pd.read_csv("sanity_check_output/sanity_check_results.csv")
flagged = coherent['n_incoherent'].fillna(0) > 0
print(f"{flagged.sum()} valeurs incohérentes entre Zea et Vel et Width")
print(f"{flagged.sum()/coherent['is_file_available'].sum() * 100:.2f}% des valeurs où Z est incohérent avec Vel et Width")
```

```python
# aberrant laplacien
laplacien = pd.read_csv("sanity_check_output/sanity_check_results.csv")
total_aberrant = laplacien['n_aberrant_laplacian'].sum()
print(f"{total_aberrant} aberrant values detected with gradient 2D")
```

```python
# spikes :
spikes = pd.read_csv("sanity_check_output/sanity_check_results.csv")
total_isolated = spikes['n_isolated'].sum()
print(f"{total_isolated} isolated detections")
flagged = spikes['n_isolated'].fillna(0) > 0
print(f"{flagged.sum()} files with at least one isolated detection")
print(f"or {flagged.sum()/spikes['is_file_available'].sum() * 100:.2f}% of the files")
```

```python
# spikes of noise : 
noise_spikes = pd.read_csv("sanity_check_output/sanity_check_results.csv")
total_isolated_noise = noise_spikes['n_isolated_weak'].sum()
flagged = noise_spikes['n_isolated_weak'].fillna(0) > 0
print(f"{total_isolated_noise} isolated detections of noise")
total_isolated_noise_files = flagged.sum()
print(f"{total_isolated_noise_files} files with at least one isolated detection of noise")
print(f"or {total_isolated_noise_files / noise_spikes['is_file_available'].sum() * 100:.2f}% of the files")
```

```python
# above echotop : 
echo_df = pd.read_csv("sanity_check_output/sanity_check_results.csv")
print(f"number of files with value above echotop : {echo_df['has_above_echotop'].sum()}")
```
